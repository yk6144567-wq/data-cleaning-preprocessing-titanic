```python
# Task 2 - Exploratory Data Analysis (EDA)
# Dataset: Iris Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()
```

```python
print("Column Names:")
print(df.columns.tolist())
```

```python
df.describe()
```

```python
df.isnull().sum()
```

Observation:
The dataset contains no missing values, so no missing-value treatment is required.

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="species")

plt.title("Distribution of Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.show()
```

Observation:
All three Iris species have an equal number of observations, with 50 samples for each species.

```python
df.hist(figsize=(10,8), bins=15)

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()
plt.show()
```

Observation:
The histograms show the distribution of the four numerical features. Petal length and petal width show more noticeable differences in their distributions compared with the sepal measurements.

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=df.drop(columns="species"))

plt.title("Boxplot of Iris Features")
plt.xticks(rotation=20)

plt.show()
```

Observation:
The boxplots help identify the spread and possible outliers in each numerical feature. Some observations in sepal width appear farther from the central range.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()
```

Observation:
Petal length and petal width have a strong relationship. The species form relatively distinct groups based on these two features.

```python
plt.figure(figsize=(8,6))

correlation = df.drop(columns="species").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

Observation:
The correlation heatmap shows that petal length and petal width have a strong positive correlation. Sepal length also has a positive relationship with the petal measurements.
# Top 3 Insights
### 1. Petal features are highly related
Petal length and petal width show a strong positive correlation.

### 2. Iris species can be separated using petal measurements
The scatter plot shows that petal length and petal width create relatively distinct groups for the three species.

### 3. The dataset is clean and balanced

# Interview Questions

### 1. How do you distinguish a genuine outlier from a data entry error?

A genuine outlier is an unusual observation that can represent a real event or natural variation. A data entry error is usually caused by incorrect data collection, typing, measurement, or formatting. We should investigate the source and context before removing an outlier.

### 2. What's the difference between correlation and causation, and why does it matter in EDA?

Correlation means that two variables are associated with each other. Causation means that a change in one variable directly causes a change in another. Correlation alone does not prove causation, so this distinction is important when interpreting EDA results.

### 3. Which chart type would you use to show a trend over 12 months, and why?

A line chart would be suitable because it clearly shows how a numerical value changes over time and makes trends easier to identify.
